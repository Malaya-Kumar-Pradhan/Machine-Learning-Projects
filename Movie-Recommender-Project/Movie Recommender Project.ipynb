```python
import pandas as pd
import numpy as np
```

```python
movies=pd.read_csv("data/tmdb-movies.csv")
```

```python
movies.head(1)
```

```python
m1=movies[["id","original_title","cast","director","overview","genres"]]
```

```python
m1.isnull().sum()
```

```python
m1.dropna(inplace=True)
```

```python
m1.duplicated().sum()
```

```python
m1.shape
```

```python
m1.drop_duplicates(keep="first",inplace=True)
```

```python
m1
```

```python
m1["cast"][1]
```

```python
m1["cast"]=m1["cast"].apply(lambda x:x.replace(" ",""))
m1["cast"]=m1["cast"].apply(lambda x:x.replace("|"," "))
```

```python
#m1["cast"]=m1["cast"].apply(lambda x:x.replace(" ",""))
m1["genres"]=m1["genres"].apply(lambda x:x.replace("|"," "))
```

```python
m1["director"]=m1["director"].apply(lambda x:x.replace(" ",""))
```

```python
m1
```

```python
m1["tags"]=m1["overview"]+m1["genres"]+m1["cast"]+m1["director"]
```

```python
m1["tags"]=m1["tags"].apply(lambda x:x.lower())
```

```python
m2=m1[["id","original_title","tags"]]
```

```python
m2.head(3)
```

```python
m2.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=9000,stop_words="english")
```

```python
vectors=cv.fit_transform(m2["tags"]).toarray()
```

```python
vectors[0]
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
m2["tags"]=m2["tags"].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[120]
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index=m2[m2["original_title"]==movie].index[0]
    distances=similarity[movie_index]
    movie_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in movie_list:
        print(m2["original_title"][i[0]])
```

```python
recommend("Inside Out")
```
